```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from wordcloud import WordCloud
import datetime
import nltk
from nltk.corpus import stopwords
import numpy as np

nltk.download('stopwords')
stop_words = set(stopwords.words('english'))

def generate_word_cloud(reviews_df, apps_df):
    health_fitness_apps = apps_df[apps_df['Category'] == 'HEALTH_AND_FITNESS']['App'].unique()
    filtered_reviews = reviews_df[(reviews_df['Sentiment'] == 'Positive') & (reviews_df['App'].isin(health_fitness_apps))]
    
    text = ' '.join(filtered_reviews['Translated_Review'].dropna())
    words = [word for word in text.split() if word.lower() not in stop_words and word.lower() not in health_fitness_apps]
    
    if words:
        wordcloud = WordCloud(width=800, height=400, background_color='white').generate(' '.join(words))
        plt.figure(figsize=(10, 5))
        plt.imshow(wordcloud, interpolation='bilinear')
        plt.axis('off')
        plt.show()
    else:
        print("No words available for word cloud generation.")

def clean_numeric_columns(apps_df):
    # Convert 'Installs' to numeric
    apps_df['Installs'] = apps_df['Installs'].astype(str).str.replace(r'[+,Free]', '', regex=True)
    apps_df['Installs'] = pd.to_numeric(apps_df['Installs'], errors='coerce').fillna(0).astype(int)

    # Convert 'Price' to numeric
    apps_df['Price'] = apps_df['Price'].astype(str).str.replace(r'[$]', '', regex=True)
    apps_df['Price'] = pd.to_numeric(apps_df['Price'], errors='coerce').fillna(0)

    # Extract 'Android Ver' as numeric
    apps_df['Android Ver'] = apps_df['Android Ver'].astype(str).str.extract(r'(\d+\.?\d*)')
    apps_df['Android Ver'] = pd.to_numeric(apps_df['Android Ver'], errors='coerce')

    # Convert 'Size' to MB (handling 'k' and 'M')
    apps_df['Size'] = apps_df['Size'].astype(str).str.replace('Varies with device', 'NaN')
    apps_df['Size'] = apps_df['Size'].astype(str).replace({'M': '', 'k': ''}, regex=True)
    apps_df['Size'] = pd.to_numeric(apps_df['Size'], errors='coerce')

    # Convert KB to MB
    apps_df['Size'] = np.where(apps_df['Size'] < 10, apps_df['Size'] * 1024, apps_df['Size'])

    # Calculate revenue
    apps_df['Revenue'] = np.where(apps_df['Price'] > 0, apps_df['Price'] * apps_df['Installs'], 0)

    return apps_df

def dual_axis_chart(apps_df):
    apps_df = clean_numeric_columns(apps_df)

    now = datetime.datetime.now()
    if not (13 <= now.hour < 14):  
        print("Dual-axis chart is only available between 1 PM to 2 PM IST.")
        return

    print("Total apps before filtering:", len(apps_df))

    filtered_apps = apps_df.copy()  # Ensure filtered_apps is defined
    filtered_apps = filtered_apps[filtered_apps['Installs'] > 10000]
    print("After Installs > 10,000:", len(filtered_apps))

    filtered_apps = filtered_apps[filtered_apps['Android Ver'] > 4.0]
    print("After Android Ver > 4.0:", len(filtered_apps))

    filtered_apps = filtered_apps[filtered_apps['Size'] > 15]
    print("After Size > 15MB:", len(filtered_apps))

    filtered_apps = filtered_apps[filtered_apps['Content Rating'] == 'Everyone']
    print("After Content Rating = Everyone:", len(filtered_apps))

    filtered_apps = filtered_apps[filtered_apps['App'].str.len() <= 30]
    print("After App Name <= 30 chars:", len(filtered_apps))

    filtered_apps = filtered_apps[filtered_apps['Revenue'] > 10000]  
    print("After Revenue > $10,000:", len(filtered_apps))

    if filtered_apps.empty:
        print("❌ No data available for the dual-axis chart.")
        return

    top_categories = filtered_apps['Category'].value_counts().nlargest(3).index
    data = filtered_apps[filtered_apps['Category'].isin(top_categories)]

    avg_installs = data.groupby(['Category', 'Type'])['Installs'].mean().unstack()
    avg_revenue = data.groupby(['Category', 'Type'])['Revenue'].mean().unstack()

    fig, ax1 = plt.subplots(figsize=(10, 5))
    ax2 = ax1.twinx()

    avg_installs.plot(kind='bar', ax=ax1, position=1, width=0.4, color=['blue', 'orange'])
    avg_revenue.plot(kind='bar', ax=ax2, position=0, width=0.4, color=['green', 'red'])

    ax1.set_ylabel('Avg Installs')
    ax2.set_ylabel('Avg Revenue')
    plt.title('Comparison of Free vs Paid Apps in Top 3 Categories')
    plt.show()

def bubble_chart(apps_df):
    apps_df = clean_numeric_columns(apps_df)
    now = datetime.datetime.now()
    if 17 <= now.hour <= 19:  
        filtered_apps = apps_df[(apps_df['Category'] == 'GAME') & 
                                (apps_df['Rating'] > 3.5) & 
                                (apps_df['Installs'] > 50000)]
        
        if filtered_apps.empty:
            print("No data available for the bubble chart.")
            return
        
        plt.figure(figsize=(10, 5))
        sns.scatterplot(x='Size', y='Rating', size='Installs', data=filtered_apps, sizes=(10, 1000), alpha=0.5)
        plt.xlabel('App Size (MB)')
        plt.ylabel('Average Rating')
        plt.title('Games Apps: Size vs. Rating (Bubble Size = Installs)')
        plt.show()
    else:
        print("Bubble chart is only available between 5 PM to 7 PM IST.")

# Example Usage:
# apps_df = pd.read_csv('Play Store Data.csv')
# reviews_df = pd.read_csv('User Reviews.csv')
# generate_word_cloud(reviews_df, apps_df)
# dual_axis_chart(apps_df)
# bubble_chart(apps_df)
```
